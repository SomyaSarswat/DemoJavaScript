```python
import  numpy as np
import pandas as pd
import matplotlib.pyplot as plt
data={'Hours': [1,2,3,4,5],
'marks':[35,40,45,55,65]
      }

df=pd.DataFrame(data)
print(df)
plt.scatter(df['Hours'],df['marks'])
plt.title("Hours vs Marks")
plt.xlabel("Hours")
plt.ylabel("Marks")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
#input data
H=np.array([[1],[2],[3],[4],[5],[6],[7],[8]])
#output data
S=np.array([35,40,45,50,55,65,75,85])
#create Model
model=LinearRegression()
model.fit(H,S)
prediction=model.predict([[9]])
print("The predicted value for 9 th hour is",prediction)
H=np.append(H,9)
S=np.append(S,prediction)
plt.plot(H,S,'o')
plt.title("Hours vs Marks")
plt.xlabel("Hours")
plt.ylabel("Marks")
plt.show()
```
